```python

# STEP 3A: REMOVE COMPLETELY EMPTY COLUMNS

print("\n========== EMPTY COLUMN CLEANING ==========")

empty_columns = df.columns[df.isnull().all()].tolist()

print("Completely empty columns:", empty_columns)

df.drop(columns=empty_columns, inplace=True)

print("Remaining rows:", df.shape[0])
print("Remaining columns:", df.shape[1])
```

```python

# STEP 3B: STANDARDIZE COLUMN NAMES

import re

def clean_column_name(column):
    column = column.strip().lower()
    column = re.sub(r"[^a-z0-9]+", "_", column)
    return column.strip("_")

df.columns = [clean_column_name(col) for col in df.columns]

print("\n========== STANDARDIZED COLUMN NAMES ==========")
print("Total columns:", len(df.columns))
print(df.columns.tolist())
```

```python

# STEP 4A: CONVERT DATE COLUMNS

print("\n========== DATE TYPE CLEANING ==========")

date_columns = [
    "order_date_dateorders",
    "shipping_date_dateorders"
]

for col in date_columns:
    df[col] = pd.to_datetime(
        df[col],
        errors="coerce"
    )

    print(f"\nColumn: {col}")
    print("Data type:", df[col].dtype)
    print("Invalid or missing dates:", df[col].isna().sum())

print("\nSample order dates:")
print(df[date_columns].head())
```

```python

# STEP 4D: INVESTIGATE ORIGINAL DATE VALUES

import pandas as pd

file_path = r"C:\Users\pns91\OneDrive\Desktop\dcsupl\DataCoSupplyChainDataset2.csv"

raw_df = pd.read_csv(
    file_path,
    low_memory=False,
    dtype=str
)

# Clean spaces in column names
raw_df.columns = raw_df.columns.str.strip()

print("Available date-related columns:")

for col in raw_df.columns:
    if "date" in col.lower():
        print(repr(col))

# Find the date columns automatically
date_columns = [
    col for col in raw_df.columns
    if "date" in col.lower()
    and ("order" in col.lower() or "shipping" in col.lower())
]

for col in date_columns:

    raw = raw_df[col].str.strip()

    # Count genuinely blank values
    blank_mask = raw.isna() | raw.eq("")

    # Test date conversion
    parsed = pd.to_datetime(
        raw.mask(blank_mask),
        errors="coerce"
    )

    # Values that exist but cannot be parsed
    invalid_mask = (~blank_mask) & parsed.isna()

    print(f"\n========== {col} ==========")
    print("Total rows:", len(raw))
    print("Blank values:", int(blank_mask.sum()))
    print(
        "Nonblank but unparseable:",
        int(invalid_mask.sum())
    )

    if invalid_mask.any():
        print("Sample invalid values:")
        print(raw[invalid_mask].head(10).to_list())
```

```python

# STEP 4E: FIX DATE FORMATS

print("\n========== DATE FORMAT CLEANING ==========")

# These columns have different date formats in the source.
# Day-first order dates use DD-MM-YYYY HH:MM.
# Shipping dates use month/day/year where applicable.

# Re-read original date values to avoid previous NaT conversions.
raw_dates = pd.read_csv(
    r"C:\Users\pns91\OneDrive\Desktop\dcsupl\DataCoSupplyChainDataset2.csv",
    usecols=[
        "order date (DateOrders)",
        "shipping date (DateOrders)"
    ],
    low_memory=False
)

# Parse order dates using the format shown in the source.
order_dates = pd.to_datetime(
    raw_dates["order date (DateOrders)"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)

# Parse shipping dates in month/day/year format.
shipping_dates = pd.to_datetime(
    raw_dates["shipping date (DateOrders)"],
    format="%m/%d/%Y %H:%M",
    errors="coerce"
)

# Replace date columns in the cleaned DataFrame.
df["order_date_dateorders"] = order_dates
df["shipping_date_dateorders"] = shipping_dates

print("Order date invalid values:",
      df["order_date_dateorders"].isna().sum())

print("Shipping date invalid values:",
      df["shipping_date_dateorders"].isna().sum())

print("\nSample corrected dates:")
print(
    df[
        ["order_date_dateorders", "shipping_date_dateorders"]
    ].head(10)
)
```

```python
import pandas as pd

file_path = r"C:\Users\pns91\OneDrive\Desktop\dcsupl\DataCoSupplyChainDataset2.csv"

df = pd.read_csv(file_path, low_memory=False)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
# Read the original date columns
raw_dates = pd.read_csv(
    file_path,
    usecols=[
        "order date (DateOrders)",
        "shipping date (DateOrders)"
    ],
    dtype=str,
    low_memory=False
)

# Convert mixed date formats
order_dates = pd.to_datetime(
    raw_dates["order date (DateOrders)"].str.strip(),
    format="mixed",
    dayfirst=False,
    errors="coerce"
)

shipping_dates = pd.to_datetime(
    raw_dates["shipping date (DateOrders)"].str.strip(),
    format="mixed",
    dayfirst=False,
    errors="coerce"
)

# Check conversion results
print("Order dates that failed:", order_dates.isna().sum())
print("Shipping dates that failed:", shipping_dates.isna().sum())

# Compare original and converted values
display(pd.DataFrame({
    "Original order date": raw_dates["order date (DateOrders)"].head(10),
    "Parsed order date": order_dates.head(10),
    "Original shipping date": raw_dates["shipping date (DateOrders)"].head(10),
    "Parsed shipping date": shipping_dates.head(10)
}))
```

```python
# Apply the corrected dates to the dataset
df["order date (DateOrders)"] = order_dates
df["shipping date (DateOrders)"] = shipping_dates

# Verify the data types
print("Order date type:", df["order date (DateOrders)"].dtype)
print("Shipping date type:", df["shipping date (DateOrders)"].dtype)

# Display sample dates
display(
    df[
        ["order date (DateOrders)", "shipping date (DateOrders)"]
    ].head(10)
)
```

```python
# 1. Check missing values before cleaning
print("Missing values before cleaning:")
print(df.isnull().sum().sort_values(ascending=False).head(10))

# 2. Remove columns that are entirely empty or unnecessary
columns_to_drop = [
    "Product Description",
    "Customer Password",
    "Customer Email"
]

df.drop(columns=columns_to_drop, inplace=True, errors="ignore")

# 3. Fill missing customer last names
df["Customer Lname"] = df["Customer Lname"].fillna("Unknown")

# 4. Verify the changes
print("\nMissing values after cleaning:")
print(df.isnull().sum().sort_values(ascending=False).head(10))

print("\nUpdated dataset shape:", df.shape)
```

```python
# Numeric data quality checks

numeric_columns = [
    "Sales",
    "Sales per customer",
    "Order Item Product Price",
    "Product Price",
    "Order Item Quantity",
    "Order Item Discount",
    "Order Item Discount Rate",
    "Order Profit Per Order",
    "Benefit per order",
    "Days for shipping (real)",
    "Days for shipment (scheduled)"
]

# Check minimum, maximum, and negative values
for col in numeric_columns:
    print(f"\n--- {col} ---")
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
    print("Negative values:", (df[col] < 0).sum())
```

```python
# 1. Check whether discount amounts exceed product prices
discount_exceeds_price = (
    df["Order Item Discount"] >
    df["Order Item Product Price"]
)

print(
    "Discount greater than product price:",
    discount_exceeds_price.sum()
)

# 2. Check discount rate against expected range
invalid_discount_rate = (
    (df["Order Item Discount Rate"] < 0) |
    (df["Order Item Discount Rate"] > 1)
)

print(
    "Invalid discount rates:",
    invalid_discount_rate.sum()
)

# 3. Calculate shipping delay
df["Shipping Delay Days"] = (
    df["Days for shipping (real)"] -
    df["Days for shipment (scheduled)"]
)

print(
    "\nOrders shipped later than scheduled:",
    (df["Shipping Delay Days"] > 0).sum()
)

print(
    "Orders shipped on schedule:",
    (df["Shipping Delay Days"] == 0).sum()
)

print(
    "Orders shipped earlier than scheduled:",
    (df["Shipping Delay Days"] < 0).sum()
)

# 4. Check the relationship with the late delivery risk flag
print("\nLate delivery risk counts:")
print(df["Late_delivery_risk"].value_counts(dropna=False).sort_index())
```

```python
# FINAL DATA QUALITY VALIDATION

# 1. Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# 2. Check primary key
print("Duplicate Order Item IDs:", df["Order Item Id"].duplicated().sum())
print("Missing Order Item IDs:", df["Order Item Id"].isna().sum())

# 3. Check missing values
print("\nColumns with missing values:")
print(df.isna().sum()[df.isna().sum() > 0])

# 4. Check date consistency
print("\nShipping dates before order dates:",
      (df["shipping date (DateOrders)"] <
       df["order date (DateOrders)"]).sum())

# 5. Validate late-delivery risk values
print("\nLate-delivery risk values:")
print(sorted(df["Late_delivery_risk"].dropna().unique()))

# 6. Check discount anomalies without deleting records
discount_issues = (
    df["Order Item Discount"] >
    df["Order Item Product Price"]
)

print("\nDiscount anomalies:", discount_issues.sum())

# 7. Confirm dataset dimensions
print("\nFinal rows:", df.shape[0])
print("Final columns:", df.shape[1])
```
